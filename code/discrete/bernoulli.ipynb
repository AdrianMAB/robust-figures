# Plot rainy events probability

## Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

## Open probabilities

```python
prob_rain=pd.read_csv("prob_rain.csv")
```

## Plot probability distribution

```python
plt.figure(figsize=(7, 5))
plt.ylabel("Probabilidad de lluvia")
plt.xlabel("Día")
plt.ylim([0,0.7])
plt.xlim([0,365])
plt.rc('font', size=14)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.plot(prob_rain,color="blue")
plt.savefig("../../generator/discrete/promedio.pdf",bbox_inches="tight")
plt.ylabel("Rainy event probability")
plt.xlabel("Day")
plt.savefig("../../generator/discrete/average.pdf",bbox_inches="tight")
plt.savefig("../../generator/discrete/average.eps",bbox_inches="tight")
plt.show()
```

## Probability distributions for all stations

```python
prob_rain_st=pd.read_csv("prob_rain_st.csv")
plt.figure(figsize=(7, 5))
plt.ylabel("Probabilidad de lluvia")
plt.xlabel("Día")
plt.ylim([0,0.8])
plt.xlim([0,365])
plt.rc('font', size=14)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.plot(prob_rain_st)
plt.savefig("../../generator/discrete/total.pdf",bbox_inches="tight")
plt.ylabel("Rainy event probability")
plt.xlabel("Day")
plt.savefig("../../generator/discrete/full.pdf",bbox_inches="tight")
plt.savefig("../../generator/discrete/full.eps",bbox_inches="tight")
plt.show()
```
